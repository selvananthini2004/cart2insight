```python
import pandas as pd
```

```python
customers=pd.read_csv("..\\data\\cleaned\\olist_customers_dataset_cleaned.csv")
Geolocation=pd.read_csv("..\\data\\cleaned\\olist_geolocation_dataset_cleaned.csv")
CustomerOrderDetails=pd.read_csv("..\\data\\cleaned\\olist_orders_dataset_cleaned.csv")
ProductPaymentDetails=pd.read_csv("..\\data\\cleaned\\olist_order_payments_dataset_cleaned.csv")
OrderReview=pd.read_csv("..\\data\\cleaned\\olist_order_reviews_dataset_cleaned.csv")
ProductCategory=pd.read_csv("..\\data\\cleaned\\product_category_name_translation_cleaned.csv")
ProductDetails=pd.read_csv("..\\data\\cleaned\\olist_products_dataset_cleaned.csv")
SellerDetails=pd.read_csv("..\\data\\cleaned\\olist_sellers_dataset_cleaned.csv")
OrderItems=pd.read_csv("..\\data\\cleaned\\olist_order_items_dataset_cleaned.csv")
```

```python
date_columns = [
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

CustomerOrderDetails[date_columns] = CustomerOrderDetails[date_columns].apply(
    pd.to_datetime
)
```

```python
date_columns = [
    "review_creation_date",
    "review_answer_timestamp"
]

OrderReview[date_columns] = OrderReview[date_columns].apply(
    pd.to_datetime
)
```

```python
OrderItems["shipping_limit_date"] = OrderItems["shipping_limit_date"].apply(pd.to_datetime)
```

<h4>Orderwise Total value</h4>

```python
OrderValue=OrderItems.groupby(["order_id"]).agg(ProductValue=("price","sum"),FreightValue=("freight_value","sum"))
```

<h4>Delivered Order</h4>

```python
DeliveredOrder=CustomerOrderDetails[CustomerOrderDetails["order_status"]=="delivered"]
DeliveredOrder
```

```python
date_columns = [
    "order_purchase_timestamp",
    "order_estimated_delivery_date"
]

DeliveredOrder[date_columns] = DeliveredOrder[date_columns].apply(
    pd.to_datetime
)
```

```python
DeliveredOrder["DeliveryDays"]=(DeliveredOrder["order_estimated_delivery_date"]-DeliveredOrder["order_purchase_timestamp"]).dt.days
DeliveredOrder
```

<h3>Delivery Delay</h3>

```python
DeliveredOrder["DeliveryDelayDays"]=(DeliveredOrder["order_delivered_customer_date"]-DeliveredOrder["order_estimated_delivery_date"]).dt.days
DeliveredOrder
```

```python
DeliveredOrder[DeliveredOrder["DeliveryDelayDays"]>0]
```

<h3>Customer Spending</h3>

```python
CustomerSpending=CustomerOrderDetails.merge(OrderItems[["order_id","price","freight_value"]],on="order_id").merge(customers[["customer_id","customer_unique_id"]], on="customer_id").groupby(["customer_unique_id"]).agg(ProductValue=("price","sum"),FreightValue=("freight_value","sum"))
```

```python
CustomerSpending["TotalSpending"]=CustomerSpending["ProductValue"]+CustomerSpending["FreightValue"]
CustomerSpending.head()
```

<h3>Average order value
</h3>

```python
CustomerSpending["AverageSpending"] = (
    CustomerSpending["TotalSpending"] /
    CustomerOrderDetails.merge(customers[["customer_id","customer_unique_id"]], on="customer_id").groupby("customer_unique_id")["order_id"].nunique()
)
CustomerSpending.head()
```

```python
CustomerSpending[CustomerSpending["AverageSpending"]!=CustomerSpending["TotalSpending"]]
```

<h3>Seller revenue</h3>

```python
SellerRevenue=SellerDetails.merge(OrderItems[["seller_id","price","freight_value"]],on="seller_id").groupby(["seller_id"]).agg(ProductValue=("price","sum"),FreightValue=("freight_value","sum"))
SellerRevenue.head()
```

```python
SellerRevenue["TotalPrice"]=SellerRevenue["ProductValue"]+SellerRevenue["FreightValue"]
SellerRevenue.head()
```
